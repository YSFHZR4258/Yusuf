```python
import csv
import datetime

# 1. Base Pizza Class (Soyut/Üst Sınıf)
class Pizza:
    def __init__(self):
        self.description = "Bilinmeyen Pizza"
        self.cost = 0.0

    def get_description(self):
        return self.description

    def get_cost(self):
        return self.cost

# 2. Somut Pizza Sınıfları (Alt Sınıflar)
class KlasikPizza(Pizza):
    def __init__(self):
        super().__init__()
        self.description = "Klasik Pizza"
        self.cost = 5.39

class MargaritaPizza(Pizza):
    def __init__(self):
        super().__init__()
        self.description = "Margarita Pizza"
        self.cost = 7.69

class TurkPizza(Pizza):
    def __init__(self):
        super().__init__()
        self.description = "Türk Pizza"
        self.cost = 4.90

class SadePizza(Pizza):
    def __init__(self):
        super().__init__()
        self.description = "Sade Pizza"
        self.cost = 6.45

class ItalyanPizza(Pizza):
    def __init__(self):
        super().__init__()
        self.description = "İtalyan Pizza"
        self.cost = 9.99

class AmerikanPizza(Pizza):
    def __init__(self):
        super().__init__()
        self.description = "Amerikan Pizza"
        self.cost = 8.86

class SpecialPizza(Pizza):
    def __init__(self):
        super().__init__()
        self.description = "Special Pizza"
        self.cost = 10.55

class EvPizza(Pizza):
    def __init__(self):
        super().__init__()
        self.description = "Ev Pizza"
        self.cost = 4.99

# 3. Decorator (Soslar için Üst Sınıf)
class Decorator(Pizza):
    def __init__(self, pizza_component):
        super().__init__()
        self.component = pizza_component

    def get_cost(self):
        return self.component.get_cost() + Pizza.get_cost(self)

    def get_description(self):
        return self.component.get_description() + ' ' + Pizza.get_description(self)

# 4. Somut Sos Sınıfları (Decorator Örnekleri)
class KasarPeyniri(Decorator):
    def __init__(self, pizza_component):
        super().__init__(pizza_component)
        self.description = "Kaşar Peyniri"
        self.cost = 2.59

    def get_cost(self):
        return self.component.get_cost() + self.cost

    def get_description(self):
        return self.component.get_description() + ", " + self.description

class RoquefortPeyniri(Decorator):
    def __init__(self, pizza_component):
        super().__init__(pizza_component)
        self.description = "Roquefort Peyniri"
        self.cost = 10.99

    def get_cost(self):
        return self.component.get_cost() + self.cost

    def get_description(self):
        return self.component.get_description() + ", " + self.description

class Zeytin(Decorator):
    def __init__(self, pizza_component):
        super().__init__(pizza_component)
        self.description = "Zeytin"
        self.cost = 5.55

    def get_cost(self):
        return self.component.get_cost() + self.cost

    def get_description(self):
        return self.component.get_description() + ", " + self.description

class Mantar(Decorator):
    def __init__(self, pizza_component):
        super().__init__(pizza_component)
        self.description = "Mantar"
        self.cost = 6.69

    def get_cost(self):
        return self.component.get_cost() + self.cost

    def get_description(self):
        return self.component.get_description() + ", " + self.description

class KeciPeyniri(Decorator):
    def __init__(self, pizza_component):
        super().__init__(pizza_component)
        self.description = "Keçi Peyniri"
        self.cost = 3.05

    def get_cost(self):
        return self.component.get_cost() + self.cost

    def get_description(self):
        return self.component.get_description() + ", " + self.description

class Et(Decorator):
    def __init__(self, pizza_component):
        super().__init__(pizza_component)
        self.description = "Et"
        self.cost = 9.80

    def get_cost(self):
        return self.component.get_cost() + self.cost

    def get_description(self):
        return self.component.get_description() + ", " + self.description

class Sogan(Decorator):
    def __init__(self, pizza_component):
        super().__init__(pizza_component)
        self.description = "Soğan"
        self.cost = 4.50

    def get_cost(self):
        return self.component.get_cost() + self.cost

    def get_description(self):
        return self.component.get_description() + ", " + self.description

class Misir(Decorator):
    def __init__(self, pizza_component):
        super().__init__(pizza_component)
        self.description = "Mısır"
        self.cost = 8.60

    def get_cost(self):
        return self.component.get_cost() + self.cost

    def get_description(self):
        return self.component.get_description() + ", " + self.description

class Biber(Decorator):
    def __init__(self, pizza_component):
        super().__init__(pizza_component)
        self.description = "Biber"
        self.cost = 2.40

    def get_cost(self):
        return self.component.get_cost() + self.cost

    def get_description(self):
        return self.component.get_description() + ", " + self.description

class Sucuk(Decorator):
    def __init__(self, pizza_component):
        super().__init__(pizza_component)
        self.description = "Sucuk"
        self.cost = 7.65

    def get_cost(self):
        return self.component.get_cost() + self.cost

    def get_description(self):
        return self.component.get_description() + ", " + self.description

# --- Örnek Sipariş Kullanımı ---
# Türk Pizzası seçip üzerine Et ve Mısır ekleyelim:
siparis = TurkPizza()
siparis = Et(siparis)
siparis = Misir(siparis)

print(f"Sipariş Detayı: {siparis.get_description()}")
print(f"Toplam Tutar: {siparis.get_cost():.2f} TL")
```

```python
import csv
import os
from datetime import datetime

class SiparisYonetici:
    def __init__(self, dosya_adi='Orders_Database.csv'):
        self.dosya_adi = dosya_adi
        self._veritabanini_hazirla()

    def _veritabanini_hazirla(self):
        """CSV dosyası yoksa sütun başlıklarıyla birlikte oluşturur."""
        if not os.path.exists(self.dosya_adi):
            with open(self.dosya_adi, mode='w', newline='', encoding='utf-8') as f:
                yazici = csv.writer(f)
                # Başlık şablonu
                yazici.writerow(['Tarih_Saat', 'Siparis_Detayi', 'Toplam_Tutar_TL'])
            print(f"{self.dosya_adi} başarıyla oluşturuldu ve hazırlandı.")

    def siparis_kaydet(self, pizza_nesnesi):
        """Bir Pizza (veya Decorator) nesnesini alıp CSV dosyasına kaydeder."""
        simdi = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
        detay = pizza_nesnesi.get_description()
        tutar = round(pizza_nesnesi.get_cost(), 2)

        with open(self.dosya_adi, mode='a', newline='', encoding='utf-8') as f:
            yazici = csv.writer(f)
            yazici.writerow([simdi, detay, tutar])

        print(f"Sipariş başarıyla veritabanına kaydedildi! ({simdi})")

    def siparisleri_listele(self):
        """Kayıtlı tüm siparişleri ekrana yazdırır."""
        print(f"\n--- {self.dosya_adi} İçindeki Kayıtlar ---")
        with open(self.dosya_adi, mode='r', encoding='utf-8') as f:
            okuyucu = csv.reader(f)
            for satir in okuyucu:
                print(" | ".join(satir))

# --- Test ve Kullanım Örneği ---
# 1. Sipariş Yöneticisini başlatalım
yonetici = SiparisYonetici('Orders_Database.csv')

# 2. Örnek bir pizza siparişi hazırlayalım (Örn: İtalyan Pizza + Mantar + Sucuk)
test_siparis = ItalyanPizza()
test_siparis = Mantar(test_siparis)
test_siparis = Sucuk(test_siparis)

# 3. Siparişi veritabanına kaydedelim
yonetici.siparis_kaydet(test_siparis)

# 4. Kaydedilen siparişleri okuyup listeleyelim
yonetici.siparisleri_listele()
```

```python
import pandas as pd
import csv

# 1. Dosya içeriğini okuyup analiz edelim
satirlar = []
with open('Orders_Database.csv', 'r', encoding='utf-8') as f:
    okuyucu = csv.reader(f)
    for satir in okuyucu:
        if satir:
            satirlar.append(satir)

# 2. Geçerli sipariş satırlarını ayıklayalım
# Geçerli bir sipariş satırı 3 sütundan oluşmalı ve son sütun (tutar) sayıya dönüştürülebilmelidir.
gecerli_siparisler = []
for satir in satirlar:
    if len(satir) == 3:
        try:
            tutar = float(satir[2])
            gecerli_siparisler.append([satir[0], satir[1], tutar])
        except ValueError:
            # Son sütun sayısal değilse (örneğin başlık satırıysa) atla
            continue

# 3. DataFrame Oluşturma
df_temiz = pd.DataFrame(gecerli_siparisler, columns=['Tarih_Saat', 'Siparis_Detayi', 'Toplam_Tutar_TL'])

# 4. Temel Analizleri Hesaplama
if not df_temiz.empty:
    toplam_siparis_sayisi = len(df_temiz)
    toplam_ciro = df_temiz['Toplam_Tutar_TL'].sum()
    ortalama_siparis = df_temiz['Toplam_Tutar_TL'].mean()

    # 5. Sonuçları Raporlama
    print("=== SİPARİŞ VERİTABANI ANALİZİ ===")
    print(f"Toplam Sipariş Sayısı : {toplam_siparis_sayisi}")
    print(f"Toplam Elde Edilen Ciro: {toplam_ciro:.2f} TL")
    print(f"Ortalama Sipariş Tutarı: {ortalama_siparis:.2f} TL\n")

    print("--- Kayıtlı Sipariş Tablosu ---")
    display(df_temiz)
else:
    print("Veritabanında analiz edilebilecek geçerli bir sipariş kaydı bulunamadı!")
```
